### Assingment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user

```python
def prime_sum(n):
    total = 0

    for j in range(2, n + 1):
        prime = True

        for i in range(2, j):
            if j % i == 0:
                prime = False
                break

        if prime:
            total = total + j

    return total


n = int(input("Enter a number: "))
print("Sum of prime numbers =", prime_sum(n))
```

### 8.9 Palindrome

```python
s1 = "hi"
s2 = "shruti"
s3 = "this"
s4 = "siht"

if s1 == s1[::-1]:
    print(s1, "is palindrome")
else:
    print(s1, "is not palindrome")

if s2 == s2[::-1]:
    print(s2, "is palindrome")
else:
    print(s2, "is not palindrome")

if s3 == s3[::-1]:
    print(s3, "is palindrome")
else:
    print(s3, "is not palindrome")

if s4 == s4[::-1]:
    print(s4, "is palindrome")
else:
    print(s4, "is not palindrome")
```
