###1. Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L (10 numbers in list now).

####i. Print L

```python
roll_no = "1024160076"
L = [int(digit) * 10 for digit in roll_no]

print("Initial L:", L)
```

```python
L.sort()
print("Ascending order:", L)


L.sort(reverse=True)
print("Descending order:", L)
```

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

vi. List comprehension for elements greater than average

```python
average = sum(L) / len(L)
new_list = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", new_list)
```

###2. Create a tuple named scores containing 8 marks: use the first 8 values from your list L in Q1.

```python


# Create tuple using first 8 values from L
scores = tuple(L[:8])

print("Scores:", scores)

# i. Highest score and its index
highest = max(scores)
highest_index = scores.index(highest)

# Lowest score and its count
lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest appears:", lowest_count)


# ii. Reverse the tuple and return as a list
# Tuples are immutable, so they cannot be reversed in place.
reversed_scores = list(scores[::-1])

print("Reversed tuple as list:", reversed_scores)


# iii. Ask user to input a score
score = int(input("Enter a score: "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score not present in tuple")


# iv. Attempt to change an element
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
    # Tuples are immutable, so their elements cannot be changed;
    # lists are mutable and allow element modification.


# v. Unpacking using *
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

###3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).

```python

import random

# Set random seed using roll number
random.seed(1024160076)

# i. Generate 100 random numbers between 100 and 900
numbers = [random.randint(100, 900) for _ in range(100)]

print("Random Numbers:")
print(numbers)


# ii. Count and print all odd numbers
odd_numbers = [num for num in numbers if num % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))


# iii. Count and print all even numbers
even_numbers = [num for num in numbers if num % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))


# iv. Count and print all prime numbers

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


prime_numbers = [num for num in numbers if is_prime(num)]

print("\nPrime numbers:")
print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))


# v. Number occurring most frequently
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency)
```

###4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

```python


# Digits of roll number
digits = [1, 0, 2, 4, 1, 6, 0, 0, 7, 6]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)


# vi. Union of A and B
union = A.union(B)
print("\nUnion of A and B:", union)


# vii. Intersection of A and B
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)


# viii. A - B and B - A separately
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives elements present only in one set,
# while symmetric_difference() gives elements present in either set but not both.


# ix. Symmetric difference
symmetric_diff = A.symmetric_difference(B)
print("Symmetric Difference:", symmetric_diff)


# x. Check subset and superset
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))


# xi. Ask user for a value X
X = int(input("\nEnter a value X: "))

A.discard(X)

print("Set A after discard:", A)

# discard() is safer because it does not give an error if X is not present;
# remove() raises a KeyError when the value does not exist.
```

###5. Consider the following dictionary:
 my_dict = { "name": "<your name>",
  "roll_no": "<your roll number>",
   "branch": "<your branch>",
   "age": <your age>,
   "city": "<your home city>" }

```python


# Original dictionary
my_dict = {
    "name": "Shruti",
    "roll_no": "1024160076",
    "branch": "Computer Science and Engineering",
    "age": 20,
    "city": "Bathinda"
}

print("Original dictionary:")
print(my_dict)


# i. Rename "city" to "location" using pop()
my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)


# ii. Add a new key "cgpa"
my_dict["cgpa"] = 8.96

print("\nAfter adding CGPA:")
print(my_dict)


# iii. Increase age by 1
my_dict["age"] = my_dict["age"] + 1

print("\nAfter increasing age by 1:")
print(my_dict)


# iv. Delete "branch" using pop() and del in two separate copies

# Method 1: Using pop()
dict_pop = my_dict.copy()
removed_value = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Value returned by pop():", removed_value)


# Method 2: Using del
dict_del = my_dict.copy()
del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop() removes the key and returns its value,
# while del removes the key but does not return its value.


# v. Iterate over dictionary using .items()
print("\nKey → Value:")

for key, value in my_dict.items():
    print(key, "→", value)


# vi. Check whether "email" exists
print("\nChecking email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present in the dictionary.")


# vii. Create a second dictionary with friend's details
friend_dict = {
    "name": "Rahul",
    "roll_no": "1024160050",
    "branch": "Computer Science and Engineering",
    "age": 21,
    "city": "Chandigarh"
}

# Merge both dictionaries
merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

# If both dictionaries have the same key,
# the value from the second dictionary (friend_dict) wins.


# viii. Dictionary comprehension
# Create a dictionary containing only string values

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:")
print(string_values)
```
