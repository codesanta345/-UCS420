Q1 Fixed entries

```python
import pandas as pd
roll_number = "1024160076"
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
categories = ["billing", "account", "general"]
last_two_digits = [int(roll_number[-2]), int(roll_number[-1])]

personalized_entries = []

for d in last_two_digits:
    category = categories[d % 3]

    if category == "account":
        entry = {
            "question": "how do i update my registered mobile number",
            "answer": "Go to Account Settings and update your registered mobile number.",
            "keywords": "mobile number update account",
            "category": category
        }

    elif category == "billing":
        entry = {
            "question": "how can i check my payment status",
            "answer": "You can check your payment status in the Billing section.",
            "keywords": "payment status billing transaction",
            "category": category
        }

    else:
        entry = {
            "question": "how can i contact customer support",
            "answer": "You can contact customer support through the Help section.",
            "keywords": "support help contact",
            "category": category
        }

    personalized_entries.append(entry)

all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print("Q1: Final 6-row DataFrame")
print(df)
```

Q2 Generate and Score a Hypothesis

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        question_words = set(row["question"].lower().split())
        keyword_words = set(row["keywords"].lower().split())

        matching_words = query_words.intersection(
            question_words.union(keyword_words)
        )

        score = len(matching_words)

        if score > 0:
            results.append((index, score))


    results.sort(key=lambda x: x[1], reverse=True)

    return results


query = "how can i pay fee"

print("\nQ2: Scoring Results")
results = score_query(query, df)

for index, score in results:
    print("Score:", score)
    print(df.loc[index])
    print()
```

Q6: Modified scoring function with tie handling


```python
def score_query_with_ties(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        question_words = set(row["question"].lower().split())
        keyword_words = set(row["keywords"].lower().split())

        matching_words = query_words.intersection(
            question_words.union(keyword_words)
        )

        score = len(matching_words)

        if score > 0:
            results.append((index, score))

    if not results:
        print("No matching entries found.")
        return

    results.sort(key=lambda x: x[1], reverse=True)

    highest_score = results[0][1]


    highest_matches = [
        (index, score)
        for index, score in results
        if score == highest_score
    ]

    if len(highest_matches) > 1:
        print("\nTie detected! All highest-scoring entries:")

        for index, score in highest_matches:
            print("\nScore:", score)
            print(df.loc[index])

    else:
        print("\nUnique highest-scoring entry:")

        index, score = highest_matches[0]

        print("Score:", score)
        print(df.loc[index])



print("\nQ6: Query producing a tie")
score_query_with_ties("fee", df)



print("\nQ6: Query without a tie")
score_query_with_ties("password", df)
```
